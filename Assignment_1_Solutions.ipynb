```python
#A1Q1
file = open ('students-marks.txt')
print ('Roll', '\t Letter Grade', '\t Grade Point') 
for marks in file:
    marks=marks.split()
    mark= (int(marks[1])+int(marks[2]))/2 + int(marks[3]) + int(marks[4])
    if mark > 100: 
        print("Invalid marks") 
    elif mark >= 80:
        print(int(marks[0]), '\t\t A+', '\t\t 4.00') 
    elif mark >= 75:
        print(int(marks[0]), '\t\t A', '\t\t 3.75')  
    elif mark >= 70:
        print(int(marks[0]), '\t\t A-', '\t\t 3.50') 
    elif mark >= 65:
        print(int(marks[0]), '\t\t B+', '\t\t 3.25') 
    elif mark>=60:
        print(int(marks[0]),'\t\tB','\t\t3.00')
    elif mark>=55:
        print(int(marks[0]),'\t\tB-','\t\t2.75')
    elif mark>=50:
        print(int(marks[0]),'\t\tC+','\t\t2.50')
    elif mark>=45:
        print(int(marks[0]),'\t\tC','\t\t2.25')
    elif mark>=40:
        print(int(marks[0]),'\t\tD','\t\t2.00')
    else:
        print(int(marks[0]), '\t\t F', '\t\t 0.0') 
```

```python

#A1Q2
file = open ('regex-sum-42.txt') 
number = []
for line in file :
    line = line.strip()
    parts = line.split()
    for part in parts:
        if part.isdigit():
            number.append(float(part))
            
print("The digits are : ", number)
print("And the sum is : ", sum(number)) 
```

```python
#A1Q3
word = []
file = open("amth.txt") 
for line in file :
    line = line.strip()
    words = line.split()
    for word_list in words:
        if word_list not in word:
            word.append(word_list) 
print(word)
word.sort()
print(word)
```

```python
#A1Q4
file=open ('mbox-short.txt')

count=0
total_confidence=0

for line in file:
    line=line.strip()
    if line.startswith('X-DSPAM-Confidence:'):
        print(line)
        words=line.split()
        confidence=float(words[1])
        total_confidence+= confidence
        count+= 1

print("Total number of lines : ", count)

if count>0:
    avg=total_confidence/count
    print("Average : ", avg)
else:
    print("NO such line is found.")
```

```python
#A1Q6

file = open('mbox-short.txt')
email_count = {}

for line in file:
    line = line.strip()
    if line.startswith('From '):
        words = line.split()
        email = words[1]
        if email in email_count:
            email_count[email] += 1
        else:
            email_count[email] = 1
print(f"\n{email_count}") 

# Find the email with the highest count
max_email = None
max_count = 0

for email, count in email_count.items() :
    if count > max_count :
        max_email = email
        max_count = count 

print(f"\n Most frequent email:\n{max_email} → {max_count} times\n")  
```
